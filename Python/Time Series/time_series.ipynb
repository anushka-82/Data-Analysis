```python
from matplotlib.dates import YearLocator 
```

```python
plt.plot(data["Month"], data["#Passengers"], label = "Passenger Count")
plt.title("Monthly pasenger count (5 Year X-Axis)")
plt.xlabel("Month")
plt.ylabel("Number of Passengers")
plt.legend()
plt.gca().xaxis.set_major_locator(YearLocator(5))

plt.grid(True)
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller
```

```python
time_series_data = data["#Passengers"]
```

```python
result = adfuller(time_series_data)
```

```python
print("ADF Statistic", result[0])
print("p-value", result[1])
print("Critical Value", result[4])
```

```python
if result[1] <= 0.05:
    print("Time Series is Stationary")
else:
    print("non-stationary")
```
